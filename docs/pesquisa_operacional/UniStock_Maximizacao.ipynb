```python
from pulp import *

# Criação do modelo
model = LpProblem("UniStock_Maximizacao", LpMaximize)

# Variáveis de decisão
x = LpVariable("Projeto_Basico", lowBound=0, cat="Integer")
y = LpVariable("Projeto_Premium", lowBound=0, cat="Integer")

# Função objetivo
model += 5000 * x + 12000 * y

# Restrições
model += 20 * x + 50 * y <= 200
model += 10 * x + 15 * y <= 80

# Resolver o problema
model.solve()

# Resultados
print("Status:", LpStatus[model.status])
print("Projetos Básicos:", x.varValue)
print("Projetos Premium:", y.varValue)
print("Lucro Máximo:", value(model.objective))
```
